## Entering data into a numpy array for calculations.

You can organize measurement data in a numpy array for processing.  For example, in the measurement of index of refraction, angle and displacement data are measured for a range of incident angles.  The incident angle measurement may have an offset that needs to be corrected for, and the displacement as well.  
With a few lines of code, the raw angle and displacement measurements can be entered, corrected for the offsets, and used to calculate the index of refraction $n$ for each angle.
Then the average $n$ and standard deviation can be calculated.

Start by importing numpy and entering the measured data and offsets.
```
import numpy as np

raw_angle = np.array([155., 163., 171., 178., 185.])
offset_angle = 148.
x = np.array([3.8, 3.9, 4.0, 4.1, 4.2])
x0 = 3.7
t = 1.8
```

Subtract the offsets from the angle and position.
Convert the angle to radians for use in later calculations.
```
theta1 = np.pi*(raw_angle - offset_angle)/180.
d = x - x0
```


Now calculate the index of refraction for each angle with
$$ n_2 = \sqrt{ \dfrac{1 - \sin^2\theta_1 }{ \left( 1 - \frac{d}{t\sin\theta_1} \right)^2 } + \sin^2\theta_1 } $$
```
n2 = np.sqrt( (1-np.sin(theta1)**2)/(1-d/(t*np.sin(theta1)))**2 + np.sin(theta1)**2 )
np.set_printoptions(precision=3)
print('Estimated indices', n2)
```
Print the results using `np.set_printoptions(precision=2)` to control the number of decimal digits to print.
For indices of refraction, 2 or 3 decimal places should suffice.

Now calculate the average of the estimated indices and their standard deviation.
The average can be determined with the `np.average(array)` function.
The function `np.std(array)` calculates the standard deviation of the values in array.
```
navg = np.average(n2)
nstd = np.std(n2)
print('Average index = ', navg ,' +/- ', nstd)
print('Average index = {:.2f} +/- {:.2f}'.format(navg,nstd))
```

````python
import numpy as np

raw_angle = np.array([155., 163., 171., 178., 185.])
offset_angle = 148.
x = np.array([3.8, 3.9, 4.0, 4.1, 4.2])
x0 = 3.7
t = 1.8
theta1 = np.pi*(raw_angle - offset_angle)/180.
d = x - x0
n2 = np.sqrt( (1-np.sin(theta1)**2)/(1-d/(t*np.sin(theta1)))**2 + np.sin(theta1)**2 )
np.set_printoptions(precision=3)
print('Estimated indices', n2)
navg = np.average(n2)
nstd = np.std(n2)
print('Average index = {:.3f} +/- {:.3f}'.format(navg,nstd))
````
